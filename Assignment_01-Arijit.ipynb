```python
branches = ("CSE", "ECE", "ME")
students = {}

n = int(input("Enter number of initial students: "))
for _ in range(n):
    roll = int(input("Enter Roll Number: "))
    name = input("Enter Name: ")
    print("Available Branches:", branches)
    branch = input("Enter Branch: ")
    marks = list(map(int, input("Enter Marks (space-separated): ").split()))
    students[roll] = {"Name": name, "Branch": branch, "Marks": marks}

print("\nAdd a new student:")
roll = int(input("Enter Roll Number: "))
name = input("Enter Name: ")
print("Available Branches:", branches)
branch = input("Enter Branch: ")
marks = list(map(int, input("Enter Marks (space-separated): ").split()))
students[roll] = {"Name": name, "Branch": branch, "Marks": marks}

del_roll = int(input("\nEnter Roll Number to delete: "))
if del_roll in students:
    del students[del_roll]
else:
    print("Student not found.")

upd_roll = int(input("\nEnter Roll Number to update: "))
if upd_roll in students:
    students[upd_roll]["Name"] = input("Enter new Name: ")
    print("Available Branches:", branches)
    students[upd_roll]["Branch"] = input("Enter new Branch: ")
    students[upd_roll]["Marks"] = list(map(int, input("Enter new Marks (space-separated): ").split()))
else:
    print("Student not found.")

print("\nFinal Student Records:")
for roll_no, details in students.items():
    print(f"Roll No: {roll_no} | Name: {details['Name']} | Branch: {details['Branch']} | Marks: {details['Marks']}")
```
